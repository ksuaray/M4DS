# **Curve Sketching Part I - Learning form the First and Second Derivatives**

## Import Libraries

```python
import sympy as sp
import numpy as np
import plotly.express as px
import pandas as pd
```

## Define Function $R(x)$

```python
# Define the symbolic variable
x = sp.Symbol('x')

# Define the function R(x)
R_x = x**3 - 6*x**2 + 9*x
print(f"R(x) = ")
R_x
```

## Sketch Function on Specified Interval
NOTE: It is important to change the interval to all real numbers if you are not restricted to a closed interval.

```python
# Convert symbolic function to a numerical function for plotting
R_num = sp.lambdify(x, R_x, 'numpy')

# Define the interval
interval_start = 0
interval_end = 4

# Generate x values for plotting
x_vals = np.linspace(interval_start, interval_end, 500)
y_vals = R_num(x_vals)

# Create a DataFrame for plotting with Plotly Express
plot_data = pd.DataFrame({'x': x_vals, 'R(x)': y_vals})

# Create the Plotly figure
fig = px.line(plot_data, x='x', y='R(x)', title=f"Sketch of R(x) = {R_x} on [0, 4]")
fig.show()
```

## Find $R'(x)$

```python

# Calculate the first derivative R'(x)
R_prime_x = sp.diff(R_x, x)
print(f"R'(x) =")
R_prime_x
```

## Critical Numbers

```python
# Find critical points by solving R'(x) = 0
critical_points = sp.solve(R_prime_x, x)
print(f"Critical points (where R'(x) = 0): {critical_points}")

# Filter critical points within the interval [0, 4] and ensure they are real numbers
critical_points_in_interval = [p for p in critical_points if p.is_real and 0 <= p <= 4]
print(f"Critical points in [0, 4]: {critical_points_in_interval}")
```

## Find $R''(x)$

```python
# Calculate the second derivative R''(x)
R_double_prime_x = sp.diff(R_prime_x, x)
print(f"R''(x) =")
R_double_prime_x
```

## Inflection Points

```python

# Find inflection points by solving R''(x) = 0
inflection_points = sp.solve(R_double_prime_x, x)
print(f"Inflection points (where R''(x) = 0): {inflection_points}")

# Filter inflection points within the interval [0, 4]
inflection_points_in_interval = [p for p in inflection_points if 0 <= p <= 4]
print(f"Inflection points in [0, 4]: {inflection_points_in_interval}")
```

## Bring it All Together

```python
# Add endpoints
fig.add_scatter(x=[interval_start], y=[R_num(interval_start)], mode='markers',
                name=f'Endpoint R({interval_start}) = {R_num(interval_start):.2f}',
                marker=dict(color='red', size=10))
fig.add_scatter(x=[interval_end], y=[R_num(interval_end)], mode='markers',
                name=f'Endpoint R({interval_end}) = {R_num(interval_end):.2f}',
                marker=dict(color='red', size=10))

# Add critical points
for cp in critical_points_in_interval:
    cp_val = float(cp)
    fig.add_scatter(x=[cp_val], y=[R_num(cp_val)], mode='markers',
                    name=f'Critical Point ({cp_val:.2f}, {R_num(cp_val):.2f})',
                    marker=dict(color='green', size=10, symbol='circle'))

# Add inflection points
for ip in inflection_points_in_interval:
    ip_val = float(ip)
    fig.add_scatter(x=[ip_val], y=[R_num(ip_val)], mode='markers',
                    name=f'Inflection Point ({ip_val:.2f}, {R_num(ip_val):.2f})',
                    marker=dict(color='purple', size=10, symbol='diamond'))

fig.update_layout(xaxis_title='x', yaxis_title='R(x)')
fig.show()
```
