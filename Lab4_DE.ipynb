## Part 3 — Where is Growth Largest?

We will:
1. Plot $F'(x)$
2. Find where $F'(x)$ is maximized

### 3.1 Plot the derivative


```python
# Numeric derivative for plotting
Fprime_num = sp.lambdify(x, F_prime, "numpy")

df["F'(x)"] = Fprime_num(x_vals)

fig = px.line(df, x="x", y="F'(x)", title="Derivative of Logistic Function F'(x)")
fig.show()
```

### 3.2 Find where $F'(x)$ is largest (symbolically)

To maximize $F'(x)$, we look for critical points of $F'(x)$, i.e. solve:

$$
\frac{d}{dx} F'(x) = 0
$$

GTW: Compute $\frac{d}{dx} F'(x)$ and solve for critical points.


```python
# GTW: Differentiate F'(x) and solve for critical points
F_double_prime = sp.diff(F_prime, x)
crit_points = sp.solve(sp.Eq(F_double_prime, 0), x)
F_double_prime_simplified = sp.simplify(F_double_prime)

F_double_prime_simplified, crit_points
```

### 3.3 Evaluate $f(0)$ and $f'(0)$

(We expect the key point to be $x=0$.)

GTW: Compute $f(0)$ and $f'(0)$.

## Part 4 — Tangent Line at the Maximum Growth Point

The tangent line at $x=0$ is:

$$
L(x) = f(0) + f'(0)(x-0)
$$

### 4.1 Build the tangent line (symbolically)


```python
# Build the tangent line at x=0
L = sp.simplify(F0 + Fprime0*(x - 0))
L
```

### 4.2 Plot logistic curve and tangent line together

```python
# Numeric tangent line for plotting
L_num = sp.lambdify(x, L, "numpy")

fig = px.line(df, x="x", y="F(x)", title="Logistic Function with Tangent Line at x=0")
fig.add_scatter(x=x_vals, y=L_num(x_vals), mode="lines", name="Tangent at x=0")
fig.show()
```
