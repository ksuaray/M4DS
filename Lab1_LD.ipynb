## 3. NumPy: fast arrays (20 min)
While for loops are effective, they are not efficient. They make the computer work extra hard to perform the desired computation, since they compute one entry at a time.

**[NumPy (Numerical Python)](https://numpy.org/)** is a free, open-source Python library for numerical computing. It is a fundamental package widely used in science, engineering, and data science, and serves as the core for the entire scientific Python ecosystem.

NumPy **arrays** let us compute with *many* numbers at once (this is huge for data science).


So far everything we've done is part of Python "out the box". We need to import additional libraries to supercharge Python's abilities. NumPy is the first library of many:

```python
import numpy as np             #When we use tools that live in numpy but not in base Python, we'll call them by using the `np.` prefix.

ar = np.array([0, 1, 2, 3, 4])  #An array is a list-like object that lives in NumPy
print('ar:', ar)
print('ar^2:', ar**2)
print('ar + 10:', ar + 10)
```

### 3.2 Distance (Location & Distance theme)
In data science (and KNN), **distance** is everything.

For two points on a line, distance is:  
$$ d(a,b) = |a-b|$$

For two points in the plane $A(x_1,y_1)$ and $B(x_2,y_2)$:

$$d(A,B) = \sqrt{(x_1-x_2)^2 + (y_1-y_2)^2} $$


```python
# 1D distances
a = np.array([0, 2, 5, 9])
b = 4
print('distances from 4:', np.abs(a - b))
```

## 4. SymPy: symbolic math (20 min)
[SymPy](https://docs.sympy.org/latest/index.html) treats symbols like real math symbols (x, y, etc.).
We'll use it constantly this semester.

From the SymPy documentation: *Symbolic computation deals with the computation of mathematical objects symbolically. This means that the mathematical objects are represented exactly, not approximately, and mathematical expressions with unevaluated variables are left in symbolic form.*

See https://docs.sympy.org/latest/tutorials/intro-tutorial/intro.html


### 4.1 Defining symbolic functions and fundamental operations

```python
import sympy as sp        #Import the SymPy library

x = sp.symbols('x')       #Declare that `x` will be a symbol we keep in ALGEBRAIC variable form
expr = x**2 - 2*x + 1     #Create an expression in terms of our symbol
expr                      #Print the expression
```

```python
# Simplify and expand
print('expanded:', sp.expand((x-1)**2))
print('factored:', sp.factor(x**2 - 1))
```

### 2 Evaluate a symbolic function on many points
This section is strictly for fun, a taste of things to come...

SymPy is symbolic, NumPy is numeric. We can connect them with `lambdify`.


```python
f = sp.sin(x)/x
f_num = sp.lambdify(x, f, 'numpy')

xs = np.linspace(-5, 5, 401)
ys = f_num(xs)

# Fix the removable issue at x=0 by setting that entry manually
ys[np.isnan(ys)] = 1.0

print('ys has shape:', ys.shape)
print('sample values:', ys[0], ys[200], ys[-1])
```

### 4.3 Quick plot
We’ll do lots of visualization this semester.


```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(xs, ys)
plt.title('f(x) = sin(x)/x')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.show()
```
