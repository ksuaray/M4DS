## 1.2 Distance function
We'll use **Euclidean distance** to measure similarity for a pair of words:
$$
 d(\mathbf{u},\mathbf{v}) = \sqrt{\sum_{i=1}^5 (u_i - v_i)^2}
$$


```python
def dist(u, v):
    """Euclidean distance between two NumPy vectors."""
    return np.linalg.norm(E[u] - E[v])

# Quick demo:
print('dist(trees, asthma) =', dist("trees", "asthma"))
```
